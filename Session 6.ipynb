```python
#Task 1 [Create Dictionary]
```

```python
playlist_price = { "bollywood hits": 100,
                   "Punjabi songs": 150,
                   "EDM songs": 180,
                   "english songs": 120,
                   "party songs": 200 }
```

```python
print(playlist_price)
```

```python
#Task 2 [Update playlist price] 
```

```python
def update_playlist_price(playlist,new_price):
    playlist_price[playlist] = new_price

update_playlist_price("bollywood hits", 500)
```

```python
print(playlist_price)
```

```python
#Task 3 [Remove playlist]
```

```python
del playlist_price["EDM songs"]

print(playlist_price)
```
